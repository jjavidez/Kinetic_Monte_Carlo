```python
import matplotlib.pyplot as plt
import numpy as np

# Cargar datos
# 'energy_output' (Col 0: Tiempo, Col 1: Energía)
data = np.loadtxt('energy_output.txt')
time = data[:, 0]
energy = data[:, 1]

plt.figure(figsize=(10, 6))
plt.plot(time, energy, color='firebrick', linewidth=1.5)

plt.title('Evolución de la Energía en el tiempo de MC', fontsize=14)
plt.xlabel('Paso de Monte Carlo (Tiempo)', fontsize=12)
plt.ylabel('Energía Total', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.loadtxt('xpol.txt')
time = data[:, 0]
px = data[:, 1]

plt.figure(figsize=(10, 6))
plt.plot(time, px, color='royalblue', label='$P_x$')

plt.title('Polarización - Componente X', fontsize=14)
plt.xlabel('Tiempo de MC', fontsize=12)
plt.ylabel('Polarización X', fontsize=12)
plt.axhline(0, color='black', linewidth=0.8, linestyle='-') # Línea de referencia en cero
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
